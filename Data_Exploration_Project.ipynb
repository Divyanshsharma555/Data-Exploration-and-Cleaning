```python
#3). a).Check Missing Values
df.isnull().sum()
```

```python
#b). Filling Missing Values
df['discount'] = df['discount'].fillna(df['discount'].mean())
df['what_customers_said'] = df['what_customers_said'].fillna("No Review")
df['seller_name'] = df['seller_name'].fillna("Unknown Seller")
df['videos'] = df['videos'].fillna("No Video")
df['seller_information'] = df['seller_information'].fillna("No Information")
df['variations'] = df['variations'].fillna("No Variations")
```

```python
df.isnull().sum()
```

```python
#4). Performance Of Basic Operations
#a). Select Columns

sample_data=df[['title','rating','discount','final_price','currency']]
sample_data.head()
```

```python
#b). Filter Rows
filtered_row=df[df['rating']>3.0]
filtered_row.head()
```

```python
#5). Remove Duplicates
#Check
df.duplicated().sum()
```

```python
#Remove
df=df.drop_duplicates()
```

```python
# Convert initial_price to string
df['initial_price'] = df['initial_price'].astype(str)

# Remove ₹ and commas from initial price column
df['initial_price'] = df['initial_price'].str.replace('₹', '', regex=False)
df['initial_price'] = df['initial_price'].str.replace(',', '', regex=False)
df['initial_price'] = df['initial_price'].str.strip()

# Convert to numeric datatype
df['initial_price'] = pd.to_numeric(df['initial_price'], errors='coerce')

# Create Quantity column beacuse not there
df['Quantity'] = 1

# Create  column name total_amount
df['total_amount'] = df['initial_price'] * df['Quantity']

df[['title', 'initial_price', 'Quantity', 'total_amount']].head()
```

```python
# 7). Saved Cleaned Data New CSV file 
df.to_csv("cleaned_dataset.csv", index=False)
```
